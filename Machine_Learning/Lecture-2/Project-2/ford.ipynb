```python
X = df.drop(columns = ['price'], axis =0)
y = df['price']
```

```python
X
```

```python
X_one_encod = pd.get_dummies(X,columns = ['model','transmission','fuelType'],drop_first = True)
```

```python
X_one_encoded = X_one_encod.astype(int)
X_one_encoded
```

```python
from sklearn.preprocessing import LabelEncoder

Xlabel = X.copy()

for col in ['model', 'transmission', 'fuelType']:
    le = LabelEncoder()
    Xlabel[col] = le.fit_transform(Xlabel[col])
```

```python
Xlabel
```

# Standardization

```python
from sklearn.preprocessing import StandardScaler

numerical_col = ['year','mileage','tax','mpg', 'engineSize']

scaler = StandardScaler()

X_one_encoded[numerical_col] = scaler.fit_transform(X_one_encoded[numerical_col])
```

```python
X_one_encoded
```

```python
Xlabel = scaler.fit_transform(Xlabel)
```

```python
Xlabel = pd.DataFrame(Xlabel, columns=X.columns)
```

```python
Xlabel
```

# Train-Test Split

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_one_encoded,
                                                 y ,
                                                 test_size=0.20,
                                                 random_state = 42
                                                 )
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_test
```

```python
r2 =r2_score(y_test,y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-((1-r2)*(n-1)) / (n - p - 1)
print("Adjusted R2: " , adjusted_r2)
```

```python
X_train,X_test,y_train,y_test = train_test_split(Xlabel,
                                                 y ,
                                                 test_size=0.20,
                                                 random_state = 42
                                                 )
```

```python
model_2 = LinearRegression()
model_2.fit(X_train,y_train)
y_pred_2 = model_2.predict(X_test)
```

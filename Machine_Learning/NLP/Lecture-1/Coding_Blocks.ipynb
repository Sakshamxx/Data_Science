```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
data = pd.read_csv("/Users/sakshamchauhan/Downloads/Machine-Learning/Machine_Learning/NLP/Lecture-1/emails.csv")
data.iloc[0,0]
```

```python
data
```

```python
def process_text(text):
    text = text.lower()[9:]
    Sentence = re.sub(r'[^a-zA-Z\s@.]+' , '',text)
    tokens = re.split(r'[,\s]+',Sentence)
    tokens = [sst.stem(w) for w in tokens if w not in sw]
    
    return ' '.join(tokens) 
```

```python
process_text(data.iloc[0,0])
```

```python
data['text'] = data['text'].apply(process_text)
```

```python
data
```

```python
X = data['text'].values
y = data['spam'].values
```

```python
cv = CountVectorizer()
```

```python
x_data = cv.fit_transform(X)
```

```python
x_data.shape
```

```python
print(x_data[0])
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
KNN = KNeighborsClassifier(5)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x_data,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
KNN.fit(X_train, y_train)
```

```python
y_pred = KNN.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:",accuracy)
```

```python
report = classification_report(y_test, y_pred)
print(report)
```

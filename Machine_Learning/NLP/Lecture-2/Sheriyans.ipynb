```python
unique_emotion = df['emotion'].unique()
emotion_number = {}
i = 0
for emo in unique_emotion:
    emotion_number[emo] = i
    i+=1
    
df['emotion'] = df['emotion'].map(emotion_number)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
df['text']
```

```python
import string
def remove_punctuation(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punctuation)
```

```python
df['text']
```

```python
def remove_num(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new
    
df['text'] = df['text'].apply(remove_num)
```

```python
df['text']
```

```python
def remove_emoji(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new
    
df['text'] = df['text'].apply(remove_emoji)
```

```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
    words = txt.split()
    cleaned = []

    for i in words:
        if i not in stop_words:
            cleaned.append(i)

    return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
doc = [
    "I Love Pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer = CountVectorizer(ngram_range= (1,3))

X = vectorizer.fit_transform(doc)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n", X.toarray())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Split RAW TEXT and LABEL together
X_train, X_test, y_train, y_test = train_test_split(
    df["text"],
    df["emotion"],
    test_size=0.2,
    random_state=42
)

print(len(X_train))
print(len(y_train))
```

# MultiNB + BoW

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Create CountVectorizer
bow_vectorizer = CountVectorizer()

# 2. Fit on training text and transform it
X_train_bow = bow_vectorizer.fit_transform(X_train)

# 3. Transform test text using the SAME vectorizer
X_test_bow = bow_vectorizer.transform(X_test)

print(X_train_bow.shape)
print(X_test_bow.shape)

# 4. Create Naive Bayes model
MNB_model = MultinomialNB()

# 5. Train
MNB_model.fit(X_train_bow, y_train)

# 6. Predict
pred_bow = MNB_model.predict(X_test_bow)

# 7. Evaluate
print("Accuracy:", accuracy_score(y_test, pred_bow))
print(classification_report(y_test, pred_bow))
print(confusion_matrix(y_test, pred_bow))
```

# MultiNB + TFIDF

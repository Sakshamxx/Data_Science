```python
import numpy as np
from scipy import stats
```

```python
sample = [172,174,168,169,171,173,175,170,169,172]
samp_mean = np.mean(sample)
samp_std= np.std(sample,ddof= 1)
n = len(sample)
mean_pop = 170
```

```python
t_stat = (samp_mean-mean_pop)/ (samp_std/np.sqrt(n))
t_stat
```

```python
p_value = 2 * (1-stats.t.cdf(abs(t_stat), df = n-1))
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha :
    print("Reject Null Hypothese")
else:
    print("Accept Null Hypothesis")
```
